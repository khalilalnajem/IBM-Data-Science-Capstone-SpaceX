# SpaceX Falcon 9 - Exploratory Data Analysis with SQL
This notebook loads the IBM capstone launch table into SQLite and answers the project questions with SQL aggregation, filtering and ranking.


```python
import pandas as pd
import sqlite3

url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/Spacex.csv'
spacex = pd.read_csv(url)
spacex.columns = [c.strip().replace(' ','_').replace('-','_') for c in spacex.columns]
con = sqlite3.connect(':memory:')
spacex.to_sql('SPACEXTBL', con, index=False, if_exists='replace')
spacex.head()
```

## Unique launch sites


```python
pd.read_sql_query('SELECT DISTINCT Launch_Site FROM SPACEXTBL', con)
```

The SQL lab identifies the launch-site names `CCAFS LC-40`, `CCAFS SLC-40`, `KSC LC-39A`, and `VAFB SLC-4E`.


## Payload carried for NASA (CRS)


```python
pd.read_sql_query('''
SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass_kg
FROM SPACEXTBL
WHERE Customer = 'NASA (CRS)'
''', con)
```

The course SQL dataset returns **45,596 kg** total payload for NASA (CRS) missions.


## Average F9 v1.1 payload


```python
pd.read_sql_query('''
SELECT AVG(PAYLOAD_MASS__KG_) AS Avg_Payload_Mass_kg
FROM SPACEXTBL
WHERE Booster_Version LIKE 'F9 v1.1%'
''', con)
```

The average payload for the F9 v1.1 booster family is approximately **2,534.7 kg** in the course dataset.


## First successful ground-pad landing


```python
pd.read_sql_query('''
SELECT MIN(Date) AS First_Ground_Landing
FROM SPACEXTBL
WHERE Landing__Outcome = 'Success (ground pad)'
''', con)
```

The first successful ground-pad landing in the SQL lab is **2015-12-22**.


## Successful drone-ship landings for 4,000-6,000 kg payloads


```python
pd.read_sql_query('''
SELECT DISTINCT Booster_Version
FROM SPACEXTBL
WHERE Landing__Outcome = 'Success (drone ship)'
  AND PAYLOAD_MASS__KG_ > 4000
  AND PAYLOAD_MASS__KG_ < 6000
''', con)
```

## Landing outcomes during the early program


```python
pd.read_sql_query('''
SELECT Landing__Outcome, COUNT(*) AS Outcome_Count
FROM SPACEXTBL
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY Landing__Outcome
ORDER BY Outcome_Count DESC
''', con)
```
