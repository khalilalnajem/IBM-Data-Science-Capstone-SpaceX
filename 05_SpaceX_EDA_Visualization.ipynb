# SpaceX Falcon 9 - EDA with Data Visualization
This notebook explores how landing success changes with flight number, payload mass, launch site, orbit and time. The source is the 90-row cleaned IBM capstone dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv'
df = pd.read_csv(url)
df['Year'] = pd.to_datetime(df['Date']).dt.year
print(df.shape)
df.head()
```

## Flight number vs launch site
Later flight numbers contain a larger share of successful landings, suggesting operational learning and reusable-booster maturity over time.


```python
sns.catplot(data=df, x='FlightNumber', y='LaunchSite', hue='Class', kind='scatter', height=4.5, aspect=1.8)
plt.xlabel('Flight number'); plt.ylabel('Launch site'); plt.show()
```

## Payload mass vs launch site
Payload mass interacts with mission profile and launch site. Success is not determined by payload alone, but successful missions become common across a broad payload range in later flights.


```python
sns.catplot(data=df, x='PayloadMass', y='LaunchSite', hue='Class', kind='scatter', height=4.5, aspect=1.8)
plt.xlabel('Payload mass (kg)'); plt.ylabel('Launch site'); plt.show()
```

## Success rate by orbit
Grouping by orbit reveals that mission profile matters. Some orbits have consistently strong outcomes, while GTO and several early LEO/ISS missions include more failures or non-attempts.


```python
orbit_success = df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)
plt.figure(figsize=(10,4))
sns.barplot(x=orbit_success.index, y=orbit_success.values)
plt.ylabel('Landing success rate'); plt.xlabel('Orbit'); plt.ylim(0,1.05); plt.xticks(rotation=45); plt.show()
```

## Yearly landing-success trend
The strongest project-level pattern is temporal: first-stage landing success improves substantially as the Falcon 9 program matures.


```python
yearly = df.groupby('Year')['Class'].mean()
plt.figure(figsize=(9,4))
plt.plot(yearly.index, yearly.values, marker='o')
plt.ylabel('Landing success rate'); plt.xlabel('Year'); plt.ylim(0,1.05); plt.grid(alpha=.25); plt.show()
```

## Launch-site summary
The cleaned dataset contains **55 launches from CCAFS SLC 40, 22 from KSC LC 39A and 13 from VAFB SLC 4E**. KSC and VAFB show particularly strong landing ratios in the later portion of the dataset.


```python
site_summary = df.groupby('LaunchSite').agg(Launches=('Class','size'), Success_Rate=('Class','mean')).sort_values('Launches', ascending=False)
site_summary
```
